```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import least_squares

def randles(params, f):
    """Impedance of a Randles circuit with a Warburg element. f is frequency in Hz."""
    Rs, Rct, Cdl, sigma = params
    w = 2 * np.pi * f                          # angular frequency
    Zw = sigma / np.sqrt(w) * (1 - 1j)         # Warburg impedance (diffusion)
    Z = Rs + 1 / (1j * w * Cdl + 1 / (Rct + Zw))
    return Z

# True values we choose for our fake electrode
true_params = [1.2, 3.5, 1e-4, 2.0]            # Rs (ohm), Rct (ohm), Cdl (F), sigma (ohm s^-0.5)

rng = np.random.default_rng(3)                 # makes the noise repeatable
f = np.logspace(5, -2, 60)                     # 100 kHz down to 0.01 Hz
Z_clean = randles(true_params, f)
noise = 0.01 * np.abs(Z_clean) * (rng.normal(size=f.size) + 1j * rng.normal(size=f.size))
Z_meas = Z_clean + noise                       # fake "measured" data with 1% noise

plt.figure(figsize=(6, 5))
plt.plot(Z_meas.real, -Z_meas.imag, "o")
plt.xlabel("Z' (ohm)")
plt.ylabel("-Z'' (ohm)")
plt.title("Synthetic Nyquist plot")
plt.gca().set_aspect("equal", adjustable="datalim")
plt.show()
```
